# Eliminación de columnas innecesarias

```python
df = df.drop(columns=["ID", "Notas"]) # Elimina columnas inecesarias
```

```python
df.head(10)
```

# Normalización de nombres de columnas 

```python
df.columns = df.columns.str.strip().str.upper().str.replace(" ",
"_").str.replace("á", "a").str.replace("é", "e")
```

```python
df.head(5)
```

```python
# Eliminar espacios y unificar mayúsculas y minúsculas
df["CIUDAD"] = df["CIUDAD"].str.strip().str.lower()

# Corregir los nombres de las ciudades
df["CIUDAD"] = df["CIUDAD"].replace({
    "bogota": "Bogotá",
    "bogotá": "Bogotá",
    "medellin": "Medellín",
    "medellín": "Medellín",
    "cali": "Cali"
})

# Verificar los resultados
print(df["CIUDAD"].value_counts(dropna=False))
```

```python
df.head(5)
```

# Tratamiento de valores nulos

```python
# Edad: media de edades válidas entre 0 y 120
media_edad = df.loc[df["EDAD"].between(0, 120), "EDAD"].mean()
df["EDAD"] = df["EDAD"].fillna(media_edad)

# Salario: mediana
df["SALARIO"] = df["SALARIO"].fillna(df["SALARIO"].median())

# Ciudad: moda
df["CIUDAD"] = df["CIUDAD"].fillna(df["CIUDAD"].mode()[0])

# Género: moda
df["GENERO"] = df["GENERO"].fillna(df["GENERO"].mode()[0])

# Mostrar resultado
df.head(10)
```

# Asegurar tipos de datos correctos

```python
# Convertir EDAD a tipo numérico
df["EDAD"] = pd.to_numeric(df["EDAD"], errors="coerce")

# Convertir FECHA a formato datetime
df["FECHA"] = pd.to_datetime(df["FECHA"], errors="coerce")

# Verificar los tipos de datos
print(df.dtypes)
```

# Filtrado de datos fuera de rango

```python
# Eliminar registros con edades fuera del rango de 0 a 120 años
df = df[df["EDAD"].between(0, 120)]

# Eliminar registros con salarios mayores a 1.000 millones
df = df[df["SALARIO"] <= 1_000_000_000]

# Reiniciar los índices
df = df.reset_index(drop=True)

# Mostrar el dataset actualizado
display(df)
```

# Conversión de variables ordinales

```python
# Convertir los valores de NIVEL a números
df["NIVEL"] = df["NIVEL"].str.strip().str.lower().map({
    "bajo": 1,
    "medio": 2,
    "alto": 3
})

# Mostrar los resultados
display(df)
```

# Codificación de variables categóricas

```python
# Convertir GENERO y CIUDAD a variables dummy
df = pd.get_dummies(
    df,
    columns=["GENERO", "CIUDAD"],
    drop_first=True,
    dtype=int
)

# Mostrar el dataset actualizado
df.head(10)
```

# Escalado de variables numéricas

```python
%pip install scikit-learn
```

```python
from sklearn.preprocessing import MinMaxScaler

# Crear el escalador
scaler = MinMaxScaler()

# Escalar las columnas EDAD e INGRESOS
df[["EDAD", "INGRESOS"]] = scaler.fit_transform(
    df[["EDAD", "INGRESOS"]]
)

# Mostrar los resultados
display(df)
```

# Eliminación de registros duplicados

```python
# Eliminar filas duplicadas
df = df.drop_duplicates()

# Reiniciar los índices
df = df.reset_index(drop=True)

# Mostrar el dataset actualizado
display(df)
```

# Creación de nuevas variables

```python
# Calcular la utilidad
df["UTILIDAD"] = df["INGRESOS"] - df["GASTOS"]

# Mostrar el dataset actualizado
display(df)
```

```python
df = df.fillna(0)

display(df)
```

# Guardado del dataset limpio

```python
# Exportar el DataFrame a un archivo CSV
df.to_csv("datos_limpios.csv", index=False)

print("Archivo datos_limpios.csv exportado correctamente.")
```
